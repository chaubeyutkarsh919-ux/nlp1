## 2. Test Face Detection on a Synthetic Face Canvas

```python
# Create synthetic face-like circle canvas to test detection
canvas = np.ones((200, 200, 3), dtype=np.uint8) * 220
cv2.circle(canvas, (100, 100), 70, (180, 180, 180), -1)
cv2.circle(canvas, (75, 80), 8, (50, 50, 50), -1)   # Left eye
cv2.circle(canvas, (125, 80), 8, (50, 50, 50), -1)  # Right eye
cv2.ellipse(canvas, (100, 130), (35, 18), 0, 0, 180, (50, 50, 50), 3) # Smile

detector = get_face_detector()
res = detector.detect_face(canvas)
print("Detector used:", res["detector_used"])
print("Face detected:", res["face_detected"])
```
