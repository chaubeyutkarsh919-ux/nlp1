## 2. Text Preprocessing Pipeline

```python
samples = [
    "I am having a wonderful day! :-) Check https://example.com",
    "This was NOT a bad film; in fact, I loved it!",
    "Customer support didn't help at all. Terrible service :("
]

for s in samples:
    print("Original: ", s)
    print("Processed:", preprocess_text(s))
    print("-" * 50)
```

## 4. Test Sample Inferences

```python
test_sentences = [
    "I am having a wonderful day!",
    "The delivery was late and the food was freezing cold.",
    "Remarkable direction and breathtaking soundtrack. Exceeded expectations!",
    "It is not a bad solution, works reasonably well."
]

for text in test_sentences:
    res = predict_sentiment(text)
    print(f"Sentence: {text}")
    print(f"-> Sentiment: {res['sentiment'].upper()} (Confidence: {res['confidence']*100:.1f}%)")
    print(f"-> Probs: {res['probabilities']}\n")
```
